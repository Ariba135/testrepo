### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
# 1. Count the total number of duplicate rows (excluding the first occurrence)
duplicate_count = df.duplicated().sum()
print(f"Total number of duplicate rows found: {duplicate_count}")

# 2. Display the first few duplicate rows to inspect their structure
if duplicate_count > 0:
    print("\n--- First 5 Duplicate Rows ---")
    print(df[df.duplicated()].head())
else:
    print("\nNo duplicate rows detected in the dataset.")
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
# 1. Remove duplicate rows (keeps the first occurrence by default)
df = df.drop_duplicates()

# 2. Verify the removal by counting duplicates again
remaining_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows remaining: {remaining_duplicates}")
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
# Count missing values for each column
missing_values = df.isnull().sum()

print("--- Missing Values Count Per Column ---")
print(missing_values)
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
# Count the missing values specifically in the RemoteWork column
missing_remotework = df['RemoteWork'].isnull().sum()

print(f"Number of missing rows in 'RemoteWork': {missing_remotework}")
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
# Display the frequency counts for each category in RemoteWork
remotework_counts = df['RemoteWork'].value_counts()

print("--- Value Counts for RemoteWork ---")
print(remotework_counts)
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
# Extract the single most frequent value (mode) from the column
remotework_majority = df['RemoteWork'].mode()[0]

print(f"The most frequent (majority) value in RemoteWork is: '{remotework_majority}'")
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
# 1. Grab the majority value string safely using .mode()[0]
remotework_majority = df['RemoteWork'].mode()[0]

# 2. Impute missing entries via safe column reassignment
df['RemoteWork'] = df['RemoteWork'].fillna(remotework_majority)

# 3. Verify the imputation worked perfectly
remaining_missing = df['RemoteWork'].isnull().sum()
print(f"Remaining missing rows in 'RemoteWork': {remaining_missing}")
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
# 1. Look for column names containing "comp" or "salary" (case-insensitive)
comp_cols = [col for col in df.columns if 'comp' in col.lower() or 'salary' in col.lower()]

print(f"Detected compensation columns: {comp_cols}\n")

# 2. Describe their statistical distribution
if comp_cols:
    print("--- Compensation Columns Distribution Summary ---")
    print(df[comp_cols].describe())
else:
    print("No columns containing 'comp' or 'salary' were found in the dataset.")
```
