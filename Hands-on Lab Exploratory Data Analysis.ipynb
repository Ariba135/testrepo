#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# --- Step 1: Clean and Prepare Columns ---
# Convert YearsCodePro to numeric (handling potential strings like 'Less than 1 year' or 'More than 50 years')
df["YearsCodePro"] = pd.to_numeric(
    df["YearsCodePro"].astype(str).str.extract(r"(\d+)")[0], errors="coerce"
)

# Map JobSat categories to numbers if it is text (e.g., 'Very dissatisfied' to 'Very satisfied')
# Skip this mapping step if your JobSat column is already numeric (1-5 or 1-10)
if df["JobSat"].dtype == "object":
    jobsat_mapping = {
        "Very dissatisfied": 1,
        "Slightly dissatisfied": 2,
        "Neither satisfied nor dissatisfied": 3,
        "Slightly satisfied": 4,
        "Very satisfied": 5,
    }
    df["JobSat_Numeric"] = df["JobSat"].map(jobsat_mapping)
else:
    df["JobSat_Numeric"] = df["JobSat"]

# Drop missing values for this specific analysis
df_clean = df.dropna(subset=["YearsCodePro", "JobSat_Numeric"])

# --- Step 2: Create Experience Ranges ---
bins = [0, 5, 10, 20, float("inf")]
labels = ["0-5 years", "5-10 years", "10-20 years", ">20 years"]
df_clean["Experience_Range"] = pd.cut(
    df_clean["YearsCodePro"], bins=bins, labels=labels, include_lowest=True
)

# --- Step 3: Calculate Median Job Satisfaction ---
summary_df = (
    df_clean.groupby("Experience_Range", observed=False)["JobSat_Numeric"]
    .median()
    .reset_index()
)
print("Median Job Satisfaction by Experience Range:")
print(summary_df)

# --- Step 4: Visualize ---
plt.figure(figsize=(8, 5))
sns.barplot(
    data=summary_df, x="Experience_Range", y="JobSat_Numeric", palette="Blues_d"
)
plt.title("Median Job Satisfaction by Professional Coding Experience")
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Median Job Satisfaction Score")
plt.ylim(0, 5.5)  # Adjust based on your satisfaction scale max
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
# --- Step 4: Visualize (Updated to remove the deprecation warning) ---
plt.figure(figsize=(8, 5))
sns.barplot(
    data=summary_df, 
    x='Experience_Range', 
    y='JobSat_Numeric', 
    hue='Experience_Range',  # Assign hue to match the x-axis variable
    palette='Blues_d', 
    legend=False            # Suppress the redundant legend box
)
plt.title('Median Job Satisfaction by Professional Coding Experience')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Job Satisfaction Score')
plt.ylim(0, 5.5)  
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
## Write your code here
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define the ideal order for categorical satisfaction levels
# (Adjust these exact strings if your dataset uses slightly different wording)
jobsat_order = [
    "Very dissatisfied",
    "Slightly dissatisfied",
    "Neither satisfied nor dissatisfied",
    "Slightly satisfied",
    "Very satisfied",
]

# Verify if values exist in the column before ordering, otherwise auto-sort
existing_categories = [
    cat for cat in jobsat_order if cat in df["JobSat"].dropna().unique()
]
order_to_use = existing_categories if existing_categories else None

# 2. Generate the Count Plot
plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    x="JobSat",
    order=order_to_use,
    hue="JobSat",  # Assign hue to avoid the palette warning
    palette="viridis",
    legend=False,  # Remove the redundant legend box
)

# 3. Format the Graph
plt.title("Overall Distribution of Job Satisfaction", fontsize=14, pad=15)
plt.xlabel("Job Satisfaction Level", fontsize=12)
plt.ylabel("Number of Respondents", fontsize=12)

# Rotate labels slightly so long category names don't overlap each other
plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()  # Ensures labels aren't cut off at the bottom
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


```python
## Write your code here
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# --- Step 1: Clean and Explode Language Data ---
df_clean = df.dropna(subset=["LanguageHaveWorkedWith", "Country"]).copy()
df_clean["Language"] = df_clean["LanguageHaveWorkedWith"].str.split(";")
df_exploded = df_clean.explode("Language")

# --- Step 2: DYNAMIC FILTERING (No manual mapping required) ---
# Automatically get the top 5 countries in your dataset with the most respondents
top_countries = df_exploded["Country"].value_counts().nlargest(5).index
df_regional = df_exploded[df_exploded["Country"].isin(top_countries)].copy()

# Rename the column visually to match your goals
df_regional["Region"] = df_regional["Country"]

# --- Step 3: Isolate Top 10 Global Programming Languages ---
top_10_langs = df_regional["Language"].value_counts().nlargest(10).index
df_regional = df_regional[df_regional["Language"].isin(top_10_langs)]

# CRITICAL FIX: Reset duplicated index tracks left behind by .explode()
df_regional = df_regional.reset_index(drop=True)

# --- Step 4: Cross-Tabulate ---
# This variable is now guaranteed to be successfully defined
regional_trends = (
    pd.crosstab(
        df_regional["Language"], df_regional["Region"], normalize="columns"
    )
    * 100
)

# --- Step 5: Visualize Regional Trends ---
plt.figure(figsize=(12, 7))
sns.heatmap(
    regional_trends,  # This will now safely render
    annot=True,
    fmt=".1f",
    cmap="YlGnBu",
    linewidths=0.5,
    cbar_kws={"label": "% of Developers Using Language"},
)

plt.title(
    "Popularity Share (%) of Programming Languages by Top Countries",
    fontsize=14,
    pad=15,
)
plt.xlabel("Country / Region", fontsize=12)
plt.ylabel("Programming Language", fontsize=12)
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# --- Step 1: Clean and Prepare Numerical Data ---
df_scatter = df.dropna(subset=["YearsCodePro", "JobSatPoints_1"]).copy()

# Convert YearsCodePro to numeric if it's stored as strings/objects
df_scatter["YearsCodePro_Numeric"] = pd.to_numeric(
    df_scatter["YearsCodePro"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce",
)

# Ensure JobSatPoints_1 is clean and numeric
df_scatter["JobSatPoints_1"] = pd.to_numeric(
    df_scatter["JobSatPoints_1"], errors="coerce"
)
df_scatter = df_scatter.dropna(subset=["YearsCodePro_Numeric", "JobSatPoints_1"])

# --- Step 2: Create a Jittered Scatter Plot ---
plt.figure(figsize=(10, 6))

# Adding a tiny bit of random noise (jitter) keeps dots from completely overlapping
sns.scatterplot(
    data=df_scatter,
    x="YearsCodePro_Numeric",
    y="JobSatPoints_1",
    alpha=0.4,  # Makes overlapping points look darker/denser
    color="teal",
    edgecolor=None,
)

# --- Step 3: Overlay a Trend Line (Optional but highly helpful) ---
# This adds a linear regression line to visually instantly spot positive or negative correlation
sns.regplot(
    data=df_scatter,
    x="YearsCodePro_Numeric",
    y="JobSatPoints_1",
    scatter=False,  # Don't duplicate the dots
    color="crimson",
    line_kws={"linewidth": 2, "label": "Trend Line"},
)

# --- Step 4: Formatting ---
plt.title(
    "Correlation: Years of Professional Experience vs. Job Satisfaction Points",
    fontsize=14,
    pad=15,
)
plt.xlabel("Years of Professional Coding Experience (YearsCodePro)", fontsize=12)
plt.ylabel("Job Satisfaction Points (JobSatPoints_1)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# --- Step 1: Clean and Prepare Columns ---
df_edu = df.dropna(subset=["EdLevel", "Employment"]).copy()


# Function to shorten long educational string values for readability
def clean_edu_labels(text):
    text = str(text)
    if "Bachelor" in text:
        return "Bachelor's"
    elif "Master" in text:
        return "Master's"
    elif "Doctoral" in text:
        return "Doctorate"
    elif "Secondary school" in text:
        return "Secondary School"
    elif "Some college" in text:
        return "Some College"
    elif "Associate degree" in text:
        return "Associate's"
    elif "Primary" in text:
        return "Primary School"
    return text[:20] + "..."  # Fallback truncation


df_edu["EdLevel_Cleaned"] = df_edu["EdLevel"].apply(clean_edu_labels)

# --- Step 2: Cross-Tabulation (Proportional Breakdowns) ---
# Normalizing by row ('index') shows the % of each employment category within that education tier
edu_xtab = (
    pd.crosstab(df_edu["EdLevel_Cleaned"], df_edu["Employment"], normalize="index")
    * 100
)

print("--- % Employment Type Breakdown by Education Level ---")
print(edu_xtab.round(1))

# --- Step 3: Visualize with a Stacked Bar Plot ---
# Stacked bar charts are ideal for displaying composition splits across groups
plt.figure(figsize=(12, 7))
edu_xtab.plot(kind="bar", stacked=True, ax=plt.gca(), cmap="viridis")

# Formatting the Chart
plt.title(
    "Employment Status Composition by Education Level", fontsize=14, pad=15
)
plt.xlabel("Education Level (EdLevel)", fontsize=12)
plt.ylabel("Percentage Share (%)", fontsize=12)
plt.xticks(rotation=30, ha="right")
plt.ylim(0, 100)

# Move legend outside the main graph area so it doesn't cover data points
plt.legend(title="Employment Type", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```
