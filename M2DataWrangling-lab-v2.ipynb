## Tasks


#### Step 1: Import the necessary module.


### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
import pandas as pd

# 1. Gather structural metrics per column
summary_df = pd.DataFrame({
    'Data Type': df.dtypes,
    'Non-Null Count': df.count(),
    'Missing Values': df.isnull().sum(),
    'Missing Percentage (%)': (df.isnull().sum() / len(df)) * 100
})

# 2. Display the sorted dataset summary
print(f"--- Dataset Summary Table (Total Rows: {len(df)}) ---")
print(summary_df.round(2))
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
# Generate basic statistics for all numerical columns
numerical_summary = df.describe()

print("--- Summary Statistics for Numerical Columns ---")
print(numerical_summary.round(2))
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# Extract and sort unique values from the Country column
unique_countries = sorted(df['Country'].dropna().unique())

print("--- Unique Entries in Country Column ---")
for country in unique_countries:
    print(f"- {country}")
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
# Strip leading/trailing spaces and force consistent Title Case
df['Country'] = df['Country'].str.strip().str.title()
df['EdLevel'] = df['EdLevel'].str.strip()

print("Basic string cleaning completed.")
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
import pandas as pd

# 1. Generate one-hot encoded columns
# dtype=int converts the True/False boolean flags into 1s and 0s
employment_encoded = pd.get_dummies(df['Employment'], prefix='Employment', dtype=int)

# 2. Join the new binary columns back to your original DataFrame
df = pd.concat([df, employment_encoded], axis=1)

# 3. Optional: Drop the original text column if you no longer need it for modeling
# df = df.drop(columns=['Employment'])

# 4. Display the newly created one-hot columns
print("--- Newly Created One-Hot Columns ---")
print(employment_encoded.head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
# 1. Calculate missing values and sort from highest to lowest
missing_counts = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

# 2. Combine into a summary DataFrame
missing_summary = pd.DataFrame({
    'Missing Count': missing_counts,
    'Percentage (%)': missing_percentage
}).sort_values(by='Missing Count', ascending=False)

# 3. Display the top 10 columns with the most missing values
print("--- Top 10 Columns with the Most Missing Values ---")
print(missing_summary.head(10).round(2))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
# 1. Calculate the robust median value
comp_median = df['ConvertedCompYearly'].median()
print(f"Calculated Median for ConvertedCompYearly: {comp_median:,.2f}")

# 2. Impute missing entries via safe column reassignment
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(comp_median)

# 3. Verify
print(f"Remaining missing values: {df['ConvertedCompYearly'].isnull().sum()}")
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# 1. Calculate the most frequent value (mode) safely as a string
remotework_mode = df['RemoteWork'].mode()[0]
print(f"The most frequent value in 'RemoteWork' is: '{remotework_mode}'")

# 2. Impute the missing values via clean column reassignment
df['RemoteWork'] = df['RemoteWork'].fillna(remotework_mode)

# 3. Verify that the missing rows have been filled
remaining_missing = df['RemoteWork'].isnull().sum()
print(f"Remaining missing values in 'RemoteWork': {remaining_missing}")
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
# 1. Double-check and extract the current minimum and maximum values of the column
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()

print(f"Original Minimum Salary: ${min_val:,.2f}")
print(f"Original Maximum Salary: ${max_val:,.2f}\n")

# 2. Apply the Min-Max Scaling formula
# We save this into a new column so you keep your original dollar values intact
df['ConvertedCompYearly_minmax'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)

# 3. Verify the transformation (Min must be exactly 0 and Max must be exactly 1)
print("--- Min-Max Scaled Verification ---")
print(f"New Minimum Value: {df['ConvertedCompYearly_minmax'].min()}")
print(f"New Maximum Value: {df['ConvertedCompYearly_minmax'].max()}")
print("\nFirst few rows of comparison:")
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_minmax']].head())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

# 1. Apply log transformation (log(x + 1)) to handle potential 0 values safely
# We save this to a new column so you don't overwrite your raw dollar values
df['ConvertedCompYearly_log'] = np.log1p(df['ConvertedCompYearly'])

# 2. Print summary statistics to show the change in distribution
print("--- Comparison of Skewness Metrics ---")
print(f"Original Data Skewness:   {df['ConvertedCompYearly'].skew():.2f}")
print(f"Log-Transformed Skewness: {df['ConvertedCompYearly_log'].skew():.2f}")

print("\nFirst few transformed values:")
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_log']].head())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
import numpy as np
import pandas as pd

# 1. Replace text strings with approximate numerical values
df['YearsCodePro_clean'] = df['YearsCodePro'].replace({
    'Less than 1 year': 0,
    'More than 50 years': 51
})

# 2. Convert the column type to float so we can safely handle numeric ranges and NaNs
df['YearsCodePro_clean'] = pd.to_numeric(df['YearsCodePro_clean'], errors='coerce')

# 3. Handle missing values (if any) by filling them with the median experience
median_years = df['YearsCodePro_clean'].median()
df['YearsCodePro_clean'] = df['YearsCodePro_clean'].fillna(median_years)
```
