The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
# Your code here
import bs4  # BeautifulSoup for parsing HTML
import pandas as pd
import requests
```

Download the webpage at the url


```python
#your code goes here
import bs4
import pandas as pd
import requests

# URL of the web page to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"

# Download the webpage content using requests
response = requests.get(url)

# Verify the download was successful
if response.status_code == 200:
    data = response.text
    print("Webpage downloaded successfully!")
else:
    print(f"Failed to download webpage. Status code: {response.status_code}")
```

Create a soup object


```python
#your code goes here
# Create a BeautifulSoup object by parsing the HTML content
soup = bs4.BeautifulSoup(data, "html.parser")
```

Scrape the `Language name` and `annual average salary`.


```python
#your code goes here
# Find the table in the parsed HTML
table = soup.find("table")

# List to store the extracted data
popular_languages = []

# Loop through rows in the table
for row in table.find_all("tr"):
    cols = row.find_all("td")

    # Extract language name and annual average salary from corresponding columns
    # Column indices correspond to the dataset structure: language name (col 1), salary (col 3)
    language_name = cols[1].getText()
    annual_average_salary = cols[3].getText()

    popular_languages.append((language_name, annual_average_salary))

# Print the extracted data
for lang, salary in popular_languages:
    print(f"{lang}: {salary}")
```
